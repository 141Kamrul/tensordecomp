```python
%matplotlib inline
import sys
sys.path.insert(0, "/home/kamrul1418/Documents/tensordecomp")
for module_name in list(sys.modules):
    if module_name == "tensordecomp" or module_name.startswith("tensordecomp."):
        del sys.modules[module_name]
import numpy as np
import tensordecomp as td

tensor = np.arange(24, dtype=float).reshape(2, 3, 4)
methods = [
    "cp", "cp_puzzle", "tucker", "tucker_puzzle",
    "hosvd", "hosvd_puzzle", "tensor_train", "tensor_train_puzzle",
]
results = td.compare_methods(tensor, methods)

for row in results:
    print(f"{row['algorithm']:<20} time={row['execution_time_ms']:.3f} ms, relative error={row['relative_error']:.6e}")
```
